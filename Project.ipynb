```python
import matplotlib.pyplot as plt
sales_by_category = df.groupby("category")["total_amount"].sum()
plt.bar(sales_by_category.index, sales_by_category.values)
plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt
orders_by_city = df["city"].value_counts()
plt.bar(orders_by_city.index, orders_by_city.values)
plt.title("Orders by City")
plt.xlabel("City")
plt.ylabel("Orders")
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt
payment_counts = df["payment_method"].value_counts()
plt.pie(payment_counts,labels = payment_counts.index ,autopct = "%1.1f%%")
plt.title("Payment counts")
plt.xlabel("Payment")
plt.ylabel = ("Counts")
plt.show()
```

```python
import matplotlib.pyplot as plt
avg_delivery = df["delivery_days"].mean()
plt.hist(df["delivery_days"], bins = 10,color = "blue",edgecolor = "black")
plt.title("Delivery Distribution")
plt.xlabel("Delivery Days")
plt.ylabel("Number of Orders")
plt.show()
```

```python
import matplotlib.pyplot as plt
rate = df["rating"].value_counts()
plt.bar(rate.index,rate.values)
plt.title("Rating Analysis")
plt.xlabel("Rating")
plt.ylabel("Number of orders")
plt.show()
```
